# Day 1 Presentation and Competitions

## Presentation

Groups will be asked to give a brief presentation (no longer than 5 minutes) of their design to the TAs.
Each group should cover the following:

1. Present your final sketches, and any other design documents that you think are important to show your design process.
2. Discuss your group's main design objectives and the key features of the design that help it achieve them.
   Be sure to describe any changes your group made as you constructed the speaker cab, or changes you recommend for future iterations.
3. Demo your speaker design and allow the TAs to measure your speaker cab's loudness.
   For this, your code must be set to the default **sine wave** with a frequency of **1 kHz** (the standard test tone).
   `BASIC_CIRCUIT_CHECK.ino` starts at 440 Hz, so change `aSin.setFreq(440);` to `aSin.setFreq(1000);` and upload the code again.

## Loudness Competition

Each group will have their speaker's loudness measured by the WEEF TAs or Ideas Clinic co-ops.
You are only allowed to change two things in the code for the loudness measurement:

1. The wave table, selected by uncommenting the table that you wish to use.
2. The frequency of the oscillator.

![Figure 1, The two settings you may change for the loudness measurement](assets/loudness_competition_settings.png)

*Figure 1, The two settings you may change for the loudness measurement*

## Design Aesthetic Competition

Each group will have their design evaluated from an aesthetic perspective by the WEEF TAs or Ideas Clinic co-ops.
One winner will be chosen from each stream.
