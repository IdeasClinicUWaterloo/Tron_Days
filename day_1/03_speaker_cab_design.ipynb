# Activity 3: Design Plate and Speaker Cab

You are to design a **baseplate** that you will mount the PCB on, along with a **speaker cab** (cabinet) that will sit on the baseplate.
You will use `BASIC_CIRCUIT_CHECK.ino` to show the feasibility of your Tron Days design and to communicate design details.
On Day 2, you will be provided with a PCB that works with the provided synth code.

## Required Materials

Ensure that you have all of the materials listed below before you begin:

| Component Name | Number of Components |
|---|---|
| Seeeduino Nano | 1 |
| USB-C Cable (for Seeeduino) | 1 |
| Breadboard | 1 |
| Amplifier | 1 |
| Jumper Wires | 6 |
| Push Button | 1 |
| Cardboard | 1 ft² / ~0.093 m² (12" × 12") |
| Scissors | 1 |
| Hot Glue Gun | 1 |
| Glue Stick | 1 |

> **Note:** the building materials will be provided once your group shows the TAs your AutoCAD drawings.

## Design Expectations

You are expected to have completed design sketches by the end of Day 1, and they must include proper dimensions.
You are also expected to submit a 2D layout of the 12" × 12" cardboard and the cut lines of your design using AutoCAD.

There will be a competition at the end of Day 1 to see which group has created the **loudest cabinet**, and a vote on which cabinet has the **best aesthetic design**.
Winners will get bragging rights, a sense of pride in their work, and everlasting fame and glory.
Note that a common side effect of winning is a general feeling of superiority over your peers.
Do not let this develop into a god complex, as it may make you lose objectivity for winning the Day 2 competitions.

The speaker dimensions are:

- 3 inch (25.4 mm) cone
- 77.8 mm × 77.8 mm × 25.49 mm

## Testing the Speaker Circuit

Start by wiring up a circuit to test the speaker, as shown in the diagram below:

![Figure 2, Speaker test circuit](assets/speaker_test_circuit.png)

*Figure 2, Speaker test circuit*

You will run a program that outputs various waves.

1. From the Tron Days files you downloaded in the [prerequisites](../prerequisites/README.md#download-the-tron-days-files), open [`code/BASIC_CIRCUIT_CHECK/BASIC_CIRCUIT_CHECK.ino`](../code/BASIC_CIRCUIT_CHECK/BASIC_CIRCUIT_CHECK.ino).
   After the IDE opens, a pop-up window may appear, prompting you to create a new folder to store your project in. Choose **OK**.
   The IDE should look like this:

   ![Figure 3, The Arduino IDE toolbar](assets/arduino_ide_toolbar.png)

   *Figure 3, The Arduino IDE toolbar*

2. Using the provided USB-C to USB-A cable, plug your Seeeduino Nano into your computer.
   An LED on the Seeeduino Nano should turn on, indicating that it is receiving power.
3. Select the board and port that you will upload your code to.
   From the **Select Board** dropdown menu, choose **Select Other Board and Port**, then choose **Seeeduino Nano**.
   Select the port that appears in the **Ports** menu, then select **OK** to close the pop-up window.

> **If you cannot select a port, you need to install a driver:**
>
> 1. In the `drivers` folder of the Tron Days files, unzip [`CP210x_Universal_Windows_Driver.zip`](../drivers/CP210x_Universal_Windows_Driver.zip).
>    If this version does not work, download the latest version from the [official Silicon Labs CP210x driver page](https://www.silabs.com/developer-tools/usb-to-uart-bridge-vcp-drivers).
> 2. Open **Device Manager**.
> 3. Click on the Seeeduino entry (check under **Other Devices**) and pick **Install drivers**.
> 4. Browse to the unzipped driver folder and select it to be installed.
> 5. If you are having trouble, ask for help.

4. It is a good idea to verify that your code compiles before uploading it to the microcontroller.
   Press the **Verify** check mark button. The output window should appear after the code finishes compiling.
5. If there are no error messages, send the code to the Seeeduino Nano using the **Upload** arrow button.
   A message should appear in the output once the upload has finished.
6. To change the wave that is output, change lines 52–55 so that a different line is uncommented, then upload the new code to the Seeeduino Nano.
   Use `Ctrl + /` to comment and uncomment lines.

> **Note:** the code initially plays a sine wave. To play a square wave, comment out line 52 and uncomment line 54, as shown below.

![Figure 4, Switching from a sine wave to a square wave](assets/basic_circuit_square_wave.png)

*Figure 4, Switching from a sine wave to a square wave*

## Sketch and Construction

1. Based on your observations of the speaker, brainstorm design objectives for a cab to house this speaker.
   As a group, choose the 3 design objectives that you believe are the most important for your design.

   | Design Objective | Description |
   |---|---|
   | #1 | |
   | #2 | |
   | #3 | |

2. Spend 10 minutes sketching a design idea for the speaker cab based on your group's design objectives.
   Each group member should sketch their own design individually. Set a timer to help your group stay on track.
3. Take turns presenting your ideas to the group. Explain how you incorporated the design objectives into your sketch.
4. As a group, choose the design you want to build.
   Consider how feasible it is to build the design using the provided materials.
   If your design requires more materials, you may ask for more cardboard, but you **must** justify why the increased material cost is required.
5. Make a scale drawing in AutoCAD of the cardboard cutouts that you will use to assemble your speaker.
    - All of your cutouts will have to fit on a 12" × 12" sheet of cardboard.
    - Your cutting lines should be object lines, and folding lines should use hidden lines.
    - You must have a title block (the size and shape of the block don't matter this time) with the name of the speaker, the names of your group members, and the scale of your drawing.
    - Add notes to communicate supplemental information, as needed.
6. Receive a speaker from a TA, then assemble your speaker cabinet using the provided cardboard and hot glue guns.
