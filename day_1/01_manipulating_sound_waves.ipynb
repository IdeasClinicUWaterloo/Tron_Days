## 1 Sound Waves: Pitch and Amplitude

In this section, you will observe how frequency and amplitude affect how a wave sounds.

**Frequency** refers to the number of times an oscillation occurs in a given period of time.
Frequency is measured in hertz (Hz), which is equivalent to the number of oscillations per second.
For example, a frequency of 20 Hz means that 20 oscillations occur per second.
We perceive frequency as **pitch**: sound waves that oscillate at higher frequencies sound higher, while lower frequencies sound lower.

**Amplitude** denotes the maximum value of an oscillation, measured from an equilibrium point.
For sound waves, amplitude is related to **volume**: the larger the amplitude of the sound wave, the louder the sound.
Volume is measured in decibels (dB), a logarithmic scale where a difference of 10 dB (one bel) is perceived as a doubling of volume.

## 2 Additive Synthesis

Delete the existing Audacity tracks to start with an empty project.
On **two different tracks**, generate two tones with the following settings:

- Sine wave with frequency **523 Hz**, amplitude of **0.5**, and duration of **2 seconds**
- Sine wave with frequency **784 Hz**, amplitude of **0.5**, and duration of **2 seconds**

Try playing each tone separately by spacing them out and listening to them one at a time.

Now place both tones on top of each other and play them to hear them at the same time.

Notice that the two sounds create a "bigger" sound when played together.
This is because of a concept called **superposition**.
When multiple tones are played at the same time, the sound you hear is the sum of each pure tone.
Mathematically, you can observe this by adding the waveform functions together in graphing software such as [Desmos](https://www.desmos.com/calculator).

The 523 Hz and 784 Hz tones complement each other to create a pleasant, harmonized sound when played together.

## 3 Beats: Sound Waves with Time-Varying Amplitudes

Delete the existing Audacity tracks. On two different tracks, generate:

- Sine wave with frequency **300 Hz**, amplitude of **1**, and duration of **2 seconds**
- Sine wave with frequency **305 Hz**, amplitude of **1**, and duration of **2 seconds**

> **Question 4:** After playing the two tones together, what do you observe that is different from the harmonized tones in the previous section?

In the first set of tones (523 Hz and 784 Hz), the frequencies are far enough apart to create harmony.
However, when two sound waves with slightly different frequencies (such as 300 Hz and 305 Hz) are played together, they create pulses called **beats**.

This phenomenon is caused by the interference of the two sound waves superimposing on each other, which gives the effect of the volume constantly being raised and lowered.

Mathematically, we have:

$$f(t) = A\sin(300 \cdot 2\pi t) + A\sin(305 \cdot 2\pi t)$$

where $A$ is the amplitude and $t$ is time.

## 4 Synthesizer Envelopes: Non-Constant Amplitude and Frequency

In the previous section, we heard a sine function with a time-varying amplitude.
We can use different functions to modify the amplitude over time, and we can also have time-varying frequencies!

### 4.1 Non-Constant Amplitude

We can replace the non-constant amplitude from the beats section,

$$2A\cos\left(\frac{\omega_1 - \omega_2}{2}t\right)\sin\left(\frac{\omega_1 + \omega_2}{2}t\right),$$

with a decaying exponential such as

$$16e^{-6t}\sin\left(\frac{\omega_1 + \omega_2}{2}t\right).$$

This makes the sine function look and sound different, since the amplitude (and consequently, the volume) of the note decreases as the note is playing.

### 4.2 Non-Constant Frequency

We can apply the same concept to the frequency of a sine function.
Recall the general equation of a sine function,

$$y = A\sin\left[\omega(x - C)\right] + D,$$

and that the period of the function is given by $\dfrac{2\pi}{\omega}$.
The angular frequency, $\omega$, has a reciprocal relationship with the period.
This means that a small period corresponds to a high frequency.
Making $\omega$ change over time results in a non-constant frequency, which causes the pitch to fluctuate.

Did you know we can create drum-like sounds with this method?
When the frequency is defined by a decaying exponential, the pitch drops sharply, which results in a drum-like sound.

Various non-constant functions can be used to define the frequency of a wave.
While a sinusoidal amplitude produces a pulsing sound, a sinusoidal frequency creates a "siren" type of sound.

## 5 Different Types of Sound Waves

Synthesizers are used to create different types of waves that have more character than pure sine waves.
They typically create sounds by generating waveforms through methods including subtractive synthesis, additive synthesis, and frequency modulation synthesis.
