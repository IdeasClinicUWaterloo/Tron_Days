## 3 Arduino IDE

The Arduino IDE is a free and open-source integrated development environment (IDE) for Arduino and Arduino-compatible boards.
Since we will be using an Arduino-compatible board rather than one directly from the manufacturer, there are a few extra steps that we will walk you through.

Please follow the steps below closely.
This will ensure that you do not spend a considerable amount of time trying to debug seemingly nonsensical technical issues.
We have already done this for you, so please learn from our suffering.

### 3.1 Download

Download the latest version of the Arduino IDE from [the Arduino software page](https://www.arduino.cc/en/software).
There are options for each operating system on the right of the page.
Once downloaded, install the Arduino IDE on your computer.

### 3.2 Configuring the Arduino IDE for the Seeeduino Nano

Open the Arduino IDE and go to **File > Preferences**.
Alternatively, use the shortcut `Ctrl + ,` on Windows or `Cmd + ,` on macOS.

![Figure 1, Opening the Preferences menu](assets/arduino_preferences_menu.png)

*Figure 1, Opening the Preferences menu*

You should then see the menu below.
In the highlighted **"Additional boards manager URLs"** field, paste the following link:

```
https://raw.githubusercontent.com/Seeed-Studio/Seeed_Platform/master/package_legacy_seeeduino_boards_index.json
```

![Figure 2, Additional boards manager URLs field](assets/additional_boards_manager_urls.png)

*Figure 2, Additional boards manager URLs field*

Press **OK** to close the Preferences menu.
Wait for the Arduino IDE to download the information from the link that you entered, then restart the IDE.

Once you reopen the Arduino IDE, open the **Boards Manager** panel.
It can be found on the left-hand side of the IDE, or opened with `Ctrl + Shift + B` on Windows or `Cmd + Shift + B` on macOS.
Search for `Seeeduino AVR` and select **Install**.

![Figure 3, Installing Seeeduino AVR from the Boards Manager](assets/seeeduino_avr_boards_manager.png)

*Figure 3, Installing Seeeduino AVR from the Boards Manager*

### 3.3 Setting Up the Mozzi Library

#### Installing the Mozzi Library

Open the **Library Manager** by clicking the library icon on the left-hand menu of the Arduino IDE.
Alternatively, use `Ctrl + Shift + I` on Windows or `Cmd + Shift + I` on macOS (the last key is a capital "i").

Type `Mozzi` into the search bar.
**Make sure that version `1.1.0` is selected**, then select **Install**.
The Library Manager defaults to the latest version, which will not work with our code.

![Figure 5, Installing Mozzi 1.1.0](assets/mozzi_library_install.png)

*Figure 5, Installing Mozzi 1.1.0*

> **Warning:** the Arduino IDE will later ask you to update all of your libraries.
> **Do not update.** Updating will undo the change you make to the Mozzi library in the next step.

#### Configuring Mozzi

You will be using a pre-designed printed circuit board (PCB) that has been made to work with Mozzi's **HIFI** (high fidelity) mode.
To use it, we need to edit one of the files that is part of the Mozzi library.
**If this step is not taken, your synthesizer will not work properly.**

These instructions are the same for all operating systems.
On both Windows and macOS, the Arduino folder is in your `Documents` folder by default.

1. Open the folder `Documents/Arduino/libraries/Mozzi`.
2. Find the file named `mozzi_config.h`.
3. Right-click `mozzi_config.h` and select **Open with**.
   On Windows, choose **Notepad**. On macOS, choose **TextEdit**.
4. Find the highlighted lines shown below.

![Figure 6, The audio mode lines in mozzi_config.h](assets/mozzi_config_audio_mode.png)

*Figure 6, The audio mode lines in mozzi_config.h*

Edit the highlighted block so that it reads exactly as follows:

````python
//#define AUDIO_MODE STANDARD
//#define AUDIO_MODE STANDARD_PLUS
#define AUDIO_MODE HIFI
````

All we are doing is commenting out the previous mode, `STANDARD_PLUS`, and uncommenting the `HIFI` mode.
This allows our Mozzi program to behave correctly with the PCB you will be using.

Save the change to `mozzi_config.h` and close the text editor.
Mozzi is now configured for Tron Days.
