## How the `NOTES` and `TIMES` Arrays Work

- The `NOTES` array contains a set of notes to be played. The `TIMES` array stores the corresponding time (in milliseconds) that each note is played for.
- The number of elements in an array is indicated by the number enclosed in square brackets.
  This number must match the number of elements enclosed in the curly brackets. **Both arrays must have the same number of elements.**
- The notes and times enclosed in the curly brackets are separated by commas.

## Programming Your Own Song

5. As a group, choose a song and instrument that you want to recreate using the synthesizer.
   In `SYNTH_SEQUENCER.ino`, edit the `NOTES_3` and `TIMES_3` arrays according to the song you would like to play.
   A quick web search can usually find the musical notes of a particular song.
   It may also be wise to choose a song with a melody that is easy to recognize (e.g., nursery rhymes like "Mary Had a Little Lamb").

   Once you have entered the new notes and times for your chosen song, don't forget to change the number of elements when initializing your arrays (see the numbers circled below).

![Figure 2, Update the number of elements in both arrays](assets/sequencer_array_size.png)

*Figure 2, Update the number of elements in both arrays*

6. If you want a sequence to repeat, set the `sequence_looping_X` variable to `true`, where `X` is the number of the sequence.
   When looping, you must set the very first note of the list to `REST` and the very first time to `0` for the loop to work properly.
7. If you want to change the type of oscillator that you are using, modify the `internal.cpp` file.
   Comment out the current oscillator and uncomment the oscillator that you want to use. The section that you can modify is shown below:

![Figure 3, Choosing the oscillator in internal.cpp](assets/sequencer_oscillator_select.png)

*Figure 3, Choosing the oscillator in internal.cpp*

> **Note:** if you get an error saying that global variables use more than 100% of dynamic memory, try reducing the number of elements in your sequences to free up space.
