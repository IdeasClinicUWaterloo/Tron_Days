# Activity 1: Assemble the Synthesizer

First, let's make sure that each button and potentiometer is properly soldered to the PCB.
The [`PCB_CONNECTION_CHECKER.ino`](../code/PCB_CONNECTION_CHECKER/PCB_CONNECTION_CHECKER.ino) code reads the values from the buttons and potentiometers and prints them to the Serial Monitor.

1. From the Tron Days files you downloaded in the [prerequisites](../prerequisites/README.md#download-the-tron-days-files), open `code/PCB_CONNECTION_CHECKER/PCB_CONNECTION_CHECKER.ino`.
   After the IDE opens, a pop-up window may appear, prompting you to create a new folder to store your project in. Choose **OK**.
2. Using the provided USB-C to USB-A cable, plug your Seeeduino Nano into your computer.
3. Select the board and port that you will upload your code to.
   From the **Select Board** dropdown menu, choose **Select Other Board and Port**, then choose **Seeeduino Nano**.
   Select the port that appears in the **Ports** menu, then select **OK** to close the pop-up window.
4. Make sure the code compiles using the **Verify** check mark button.
   If there are no error messages, send the code to the synthesizer using the **Upload** arrow button.
   A message should appear in the output once the upload has finished.
5. Open the Serial Monitor using `Ctrl + Shift + M` (Windows) or `Cmd + Shift + M` (macOS).
   Try turning the dial on each potentiometer and pressing each button.

The value associated with each of the first five potentiometers should **increase** as the dial is turned clockwise and **decrease** as the dial is turned counterclockwise.
The button states should be `0` (low) when pressed and `1` (high) when not pressed.

It might be a good idea to add a longer delay after the print commands so that you can read the values shown on the Serial Monitor, for example `delay(3000);`.

> If your PCB is not behaving as expected, consult a member of the teaching team before proceeding to the next activity.
