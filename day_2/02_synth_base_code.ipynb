## Changing a Pin with `#define`

The `#define` directive is used to name a value that is constant (i.e., it will not change while the program is running).
When the name is used in code, the compiler replaces the name with its associated value.

Change the line `#define ADSR_NOTE_0 7` to `#define ADSR_NOTE_0 2`. Save, verify, and upload your code to the Seeeduino Nano.

## Adding More Notes

Currently, the note that can be played in ADSR mode (by pressing one of the buttons) is set up using the following lines of code.

**A.** A `#define` statement that defines the digital pin number (see line 22 in the provided code file):

```python
#define ADSR_NOTE_0 7
```

**B.** The declaration of a `bool` variable to store the current state of the button, set to `HIGH` to start (see line 39):

```python
bool notePinState0 = HIGH;
```

**C.** A `pinMode` statement in the `setup` function. It defines the `ADSR_NOTE` digital pin (D3) as an input (see line 60):

```python
pinMode(ADSR_NOTE_0, INPUT_PULLUP);
```

**D.** A `case` in the `checkInputs` function that uses `digitalRead` to read the current state of digital pin 3 (D3) (see lines 91 and 92):

```python
case 0:
  notePinState0 = digitalRead(ADSR_NOTE_0);
```

> **Note:** when adding additional notes, you will have to make another case (i.e., `case 1`, `case 2`, `case 3`, etc.).

**E.** A call to the `updateADSR` function, which plays notes based on the recorded input of all the buttons.
This function can update the ADSR mode based on up to 6 pin states (see line 122):

```python
updateADSR(notePinState0);
```

For example, if you want to use three pin states, you would make the following change:

```python
updateADSR(notePinState0, notePinState1, notePinState2);
```
